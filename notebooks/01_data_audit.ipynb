## Section 4: Product Group Audit

```python
print("ALL PRODUCT GROUPS BY COUNT:")
print("-" * 70)
for i, (group, count) in enumerate(df['Product Group'].value_counts().items(), 1):
    pct = (count / len(df)) * 100
    print(f"{i:2d}. {group:40s} | {count:6,} rows | {pct:4.1f}%")
```

```python
print("ALL PRODUCT GROUPS (alphabetical - to spot duplicates):")
print("-" * 60)
for group in sorted(df['Product Group'].dropna().unique()):
    print(f"  '{group}'")
```

```python
pg = df['Product Group'].dropna().astype(str)
space_issues = (pg != pg.str.strip()).sum()
print(f"Product Group values with trailing spaces: {space_issues:,}")

prod = df['Product'].dropna().astype(str)
prod_space = (prod != prod.str.strip()).sum()
print(f"Product values with trailing spaces:       {prod_space:,}")
```

### Findings: Product Group Audit

- 38 unique product groups found
- Inconsistent capitalization: mix of UPPER CASE and Title Case
- Trailing spaces found in 168,864 Product Group values (22% of data)
- Trailing spaces found in 1,216 Product values
- Duplicate categories found:
    - PERSONAL CARE and PERSONAL_CARE are the same category
    - CANNED AND PACK GOODS and CANNED AND PACK ITEM are the same category
- All issues will be fixed in 02_data_cleaning.ipynb
