## Section 4: Product Group Audit

```python
print("ALL PRODUCT GROUPS BY COUNT:")
print("-" * 70)
for i, (group, count) in enumerate(df['Product Group'].value_counts().items(), 1):
    pct = (count / len(df)) * 100
    print(f"{i:2d}. {group:40s} | {count:6,} rows | {pct:4.1f}%")
```
