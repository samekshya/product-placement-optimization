## Section 4: Product Group Audit

```python
print("ALL PRODUCT GROUPS BY COUNT:")
print("-" * 70)
for i, (group, count) in enumerate(df['Product Group'].value_counts().items(), 1):
    pct = (count / len(df)) * 100
    print(f"{i:2d}. {group:40s} | {count:6,} rows | {pct:4.1f}%")
```

```python
print("ALL PRODUCT GROUPS (alphabetical - to spot duplicates):")
print("-" * 60)
for group in sorted(df['Product Group'].dropna().unique()):
    print(f"  '{group}'")
```

```python
pg = df['Product Group'].dropna().astype(str)
space_issues = (pg != pg.str.strip()).sum()
print(f"Product Group values with trailing spaces: {space_issues:,}")

prod = df['Product'].dropna().astype(str)
prod_space = (prod != prod.str.strip()).sum()
print(f"Product values with trailing spaces:       {prod_space:,}")
```

## Section 6: Discount, VAT and Round Off Audit

### Findings: Discount, VAT and Round Off Audit

  - Has no analytical value
  - Decision: DROP in cleaning
- Both DISCOUNT and ROUND OFF have 158 missing values
  - Same 158 rows are missing across all three columns
  - These are likely blank or corrupted rows
  - Decision: investigate in missing values section

## Section 8: Invoice and Basket Level Audit

```python
invoice_summary = df.groupby('Inv.No').agg(
    basket_value=('Total Amount', 'sum'),
    unique_products=('Product', 'nunique'),
    total_quantity=('Qty', 'sum'),
    line_count=('Product', 'count')
).reset_index()

print("INVOICE / BASKET LEVEL AUDIT")
print("-" * 60)
print(f"Total invoices:                     {invoice_summary['Inv.No'].nunique():,}")
print(f"Average basket value:               Rs {invoice_summary['basket_value'].mean():.2f}")
print(f"Median basket value:                Rs {invoice_summary['basket_value'].median():.2f}")
print(f"Average unique products per basket: {invoice_summary['unique_products'].mean():.2f}")
print(f"Single product baskets:             {(invoice_summary['unique_products'] == 1).sum():,}")
print(f"Maximum products in one basket:     {invoice_summary['unique_products'].max():,}")
```

```python
sample_invoice = df['Inv.No'].dropna().iloc[1]
print(f"Sample invoice: {sample_invoice}")
df[df['Inv.No'] == sample_invoice]
```

### Findings: Invoice and Basket Level Audit

- 218,037 unique shopping trips confirmed
- Average basket value: Rs 1,001.59 - this is our PRIMARY KPI baseline
- Median basket value: Rs 500.00 - most customers make modest purchases
- Average 3.51 unique products per basket - sufficient for association analysis
- 75,168 single product baskets (34.5%) - these cannot be used in association rules
- Maximum 123 products in one basket - likely a bulk business purchase
- Inv.No confirmed as valid basket identifier
