# 01 - Data Audit
## Product Placement Optimization - Nepali Grocery Store

### Purpose
This notebook performs a complete audit of the raw transaction dataset 
obtained from a Nepali grocery retail store. The goal is to understand 
the dataset structure, confirm data completeness, and identify quality 
issues before any cleaning or analysis begins.

### Dataset
- **Source:** Real POS transaction data from a Nepali grocery retail store
- **Size:** 768,222 rows x 14 columns
- **Period:** July 2025 to May 2026 (10 months)

```python
# Importing necessary libraries
import pandas as pd
```

```python
#Loading raw data
df = pd.read_excel(r'D:\softwarica\Sem 6\Individual Project\product-placement-optimization\data\raw\SALES_REGISTER.csv', header=7)
print(f"Data loaded!, Structure: {len(df)} rows and {len(df.columns)} columns.")
```

```python
#First look
print(f"Columns: {list(df.columns)}")
print(f"\nFirst 5 rows:")
df.head()
```

```python
# Key numbers exploration
print("KEY NUMBERS ")
print("-" * 50)
print(f"Total unique invoices: {df['Inv.No'].nunique():,}")
print(f"Total unique products: {df['Product'].nunique():,}")
print(f"Unique product groups: {df['Product Group'].nunique():,}")
print(f"Unique customers:      {df['Customer'].nunique():,}")
print(f"Date range:            {df['Date'].min()} to {df['Date'].max()}")
```

```python
# Product groups analysis
print("ALL PRODUCT GROUPS:")
print("-" * 70)
for i, (group,count) in enumerate(df["Product Group"].value_counts().items(),1):
    pct = (count / len(df)) * 100
    print(f"{i:2d} {group:40s} | {count:10} rows | {pct:5.1f}% |")
```

```python
# Top 30 products
print("TOP 30 MOST SOLD PRODUCTS:")
print("-" * 70)
for i, (products,count) in enumerate(df['Product'].value_counts().head(30).items(),1):
    print(f"{i:2d}. {products:40s} | sold{count:>7,} times |")
```

```python
# Missing values check
print("\n1. MISSING VALUES PER COLUMN:")
nulls = df.isnull().sum()
for col, n in nulls.items():
    status = "PROBLEM" if n>0 else "OK"
    print(f"{col:25s} {n:3d} missing | {status}")
```

```python
# Duplicates check
dupes = df.duplicated().sum()
print(f"\n2. DUPLICATE ROWS: {dupes:,}")
```

```python
# Negative values
neg_qty = (df['Qty'] <= 0).sum()
neg_amount = (df['Total Amount'] <= 0).sum()
print(f"\n3. ROWS WITH QTY <= 0:          {neg_qty:,}")
print(f"   ROWS WITH TOTAL AMOUNT <= 0: {neg_amount:,}")
```

```python
# NaN in product group
nan_pg = df['Product Group'].isna().sum()
print(f"\n4. ROWS WITH NO PRODUCT GROUP:  {nan_pg:,}")
```
